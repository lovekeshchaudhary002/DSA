# Implementation of Queue using Singly Linked List

```python
# Define a Node class to create node(item,next)
class Node:
  def __init__(self,item=None,next=None):
    self.item=item
    self.next=next

# Define a class Queue to implement queue using singly linked list
# Define __init__() method to initialize front and rear and item_count reference variables
class Queue:
  def __init__(self):
    self.front=None
    self.rear=None
    self.item_count=0
  
  # Define is_empty() method to check if Queue is empty
  def is_empty(self):
    return self.item_count==0

  # Define enqueue() method to insert elements in Queue (first in)
  def enqueue(self,element):
    n=Node(element)
    if self.is_empty():
      self.front=n
      self.rear=n
    else:
      self.rear.next=n
      self.rear=n
    self.item_count+=1

  # Define dequeue() method to remove front element from Queue (first out)
  def dequeue(self):
    if self.is_empty():
      print('Queue is empty')
    elif self.front == self.rear:
      self.front=None
      self.rear=None
    else:
      self.front=self.front.next
    self.item_count-=1

  # Define get_front() method to show first element
  def get_front(self):
    if self.is_empty():
      print('Queue is empty')
    else:
      return 'front element',self.front.item

  # Define get_rear() method to show last element
  def get_rear(self):
    if self.is_empty():
      print('Queue is empty')
    else:
      return 'rear element',self.rear.item

  # Define size() method to find the number of elements in Queue
  def size(self):
    return 'size',self.item_count
```

```python
# Create object of Queue class
obj=Queue()
```

```python
# insert elements
obj.enqueue(10)
obj.enqueue(20)
obj.enqueue(30)
obj.enqueue(40)
obj.enqueue(50)
```

```python
# Show front element 
obj.get_front()
```

```python
# Show rear element
obj.get_rear()
```

```python
# Remove front element (first out)
obj.dequeue()
```

```python
# Number of elements in Queue
obj.size()
```
