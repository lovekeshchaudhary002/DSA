# Implementation of Queue using List

```python
# Define a class Queue to implement queue using List
# Define __init__() to create empty list
class Queue:
  def __init__(self):
    self.items=[]

  # Define is_empty() method to check is Queue is empty
  def is_empty(self):
    return len(self.items)==0

  # Define enqueue() method to insert element in Queue
  def enqueue(self,element):
    self.items.append(element)

  # Define dequeue() method to remove first element from Queue
  def dequeue(self):
    if self.is_empty():
      pass
    else:
      return 'Removed first element :',self.items.pop(0)

  # Define get_front() method to show first element
  def get_front(self):
    if self.is_empty():
      print('Queue is empty')
    else:
      return 'First element in Queue:',self.items[0]

  # Define get_rear() method to show last element
  def get_rear(self):
    if self.is_empty():
      print('Queue is empty')
    else:
      return 'Last element in Queue:',self.items[-1]

  # Define size() method to show size of Queue
  def size(self):
    return 'Size of Queue:',len(self.items)
```

```python
# Create object of Queue class
obj=Queue()
```

```python
# Insert elements in Queue
obj.enqueue(10)
obj.enqueue(20)
obj.enqueue(30)
obj.enqueue(40)
obj.enqueue(50)
```

```python
# Show front element in Queue
obj.get_front()
```

```python
# Show rear element in Queue
obj.get_rear()
```

```python
# Remove first element from Queue
obj.dequeue()
```

```python
# new front element
obj.get_front()
```

```python
# Total number of elements in Queue
obj.size()
```
