# Stack Using List

```python
# Define a class Stack to implement stack using List
# Create empty list
class Stack:
  def __init__(self):
    self.items=[]

  # def is_empty() method to check if stack is empty
  def is_empty(self):
    return len(self.items)==0
  
  # Define push() method to insert elements
  def push(self,element):
    self.items.append(element)

  # Define pop() method to remove top element
  def pop(self):
    if self.is_empty():
      pass
    else:
      return 'Removed element:-',self.items.pop()

  # Define peek() method to show top element
  def peek(self):
    if self.is_empty():
      print('Stack is empty')
    else:
      return 'Top element:-',self.items[-1]

  # Define size() method to find number of elements in stack
  def size(self):
    return 'Total elements in Stack:-',len(self.items)
```

```python
# Create object of Stack class
obj=Stack()
```

```python
# insert elements
obj.push(10)
obj.push(20)
obj.push(30)
obj.push(40)
obj.push(50)
```

```python
# Top element
obj.peek()
```

```python
# Remove top element
obj.pop()
```

```python
# new top element
obj.peek()
```

```python
# Size of stack
obj.size()
```
