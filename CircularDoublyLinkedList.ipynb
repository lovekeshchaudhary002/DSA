```python
# Define a class Node to describe node (prev,item,next)
class Node:
  def __init__(self,prev=None,item=None,next=None):
    self.prev=prev
    self.item=item
    self.next=next

# Define a class CDLL to initialize start reference
class CDLL:
  def __init__(self,start=None):
    self.start=start

  # Define is_empty() method in CDLL to check if list is empty
  def is_empty(self):
    return self.start==None

  # Define insert_at_start() method in CDLL to insert an element at start of list
  def insert_at_start(self,element):
    n=Node(item=element)
    if self.is_empty():
      n.next=n
      n.prev=n
      self.start=n
    else:
      n.next=self.start
      n.prev=self.start.prev
      self.start.prev.next=n
      self.start.prev=n
      self.start=n

  # Define insert_at_last() method in CDLL to insert an element at end of list
  def insert_at_last(self,element):
    n=Node(item=element)
    if self.is_empty():
      n.next=n
      n.prev=n
      self.start=n
    else:
      n.next=self.start
      n.prev=self.start.prev
      self.start.prev.next=n
      self.start.prev=n

  # Define search() method in CDLL to find a specific element in list
  def search(self,value):
    if self.is_empty():
      return None
    else:
      temp=self.start
      while temp.next is not self.start:
        if temp.item==value:
          return temp
        temp=temp.next
      if temp.item==value:
        return temp
      return None

  # Define insert_after() method in CDLL to insert an element after specific element
  def insert_after(self,temp,element):
    n=Node(item=element)
    n.next=temp.next
    n.prev=temp
    temp.next.prev=n
    temp.next=n

  # Define print_list() method in CDLL to print all elements of list
  def print_list(self):
    if self.is_empty():
      print('List is empty')
    else:
      temp=self.start
      while temp.next is not self.start:
        print(temp.item,end=' ')
        temp=temp.next
      print(temp.item)

  # Define delete_first() method in CDLL to delete first element
  def delete_first(self):
    if self.is_empty():
      pass
    if self.start.next==self.start:
      self.start=None
    else:
      self.start.prev.next=self.start.next
      self.start.next.prev=self.start.prev
      self.start=self.start.next

  # Define delete_last() method in CDLL to delete last element
  def delete_last(self):
    if self.is_empty():
      pass
    elif self.start.next==self.start:
      self.start=None
    else:
      self.start.prev.prev.next=self.start
      self.start.prev=self.start.prev.prev

  # Define delete_item() method in CDLL to delete a specific element
  def delete_item(self,value):
    if self.is_empty():
      pass
    elif self.start.next==self.start:
      self.start=None
    else:
      temp=self.start
      if temp.item==value:
        temp.next.prev=temp.prev
        temp.prev.next=temp.next
        self.start=temp.next
      else:
        while temp.next is not self.start:
          if temp.item==value:
            temp.prev.next=temp.next
            temp.next.prev=temp.prev
          temp=temp.next
          if temp.item==value:
            temp.prev.next=temp.next
            temp.next.prev=temp.prev
```

```python
# Create object of CDLL
obj=CDLL()
```

```python
# Insert elements
obj.insert_at_start(20)
obj.insert_at_start(15)
obj.insert_at_start(5)
obj.insert_at_last(30)
obj.insert_at_last(35)
obj.insert_after(obj.search(5),10)
obj.insert_after(obj.search(20),25)
obj.insert_after(obj.search(35),40)
```

```python
# Print elements of List
obj.print_list()
```

```python
# Delete elements
obj.delete_first()
obj.delete_last()
obj.delete_item(20)
```
