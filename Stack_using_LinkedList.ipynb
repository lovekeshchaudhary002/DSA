# Stack Using Singly Linked List

```python
# Define Node class to describe node (item,next)
class Node:
  def __init__(self,item=None,next=None):
    self.item=item
    self.next=next

# Define Stack class to implement stack
# start (top) reference variable and item_count variable
class Stack:
  def __init__(self,top=None):
    self.top=top
    self.item_count=0

  # Define is_empty() method to check if stack is empty
  def is_empty(self):
    return self.top==None

  # Define push method to insert elements in stack
  def push(self,element):
    n=Node(element,self.top)
    self.top=n
    self.item_count+=1

  # Define pop() method to remove top(first) element from stack
  def pop(self):
    if self.is_empty():
      pass
    else:
      top_element=self.top.item
      self.top=self.top.next
      self.item_count -= 1
      return 'Removed element:-',top_element
      

  # Define peek() method to show top element of stack
  def peek(self):
    if self.is_empty():
      print('Stack is empty')
    else:
      return 'Top element:-',self.top.item

  # Define size() method to show size of stack
  def size(self):
    return 'Total elements in stack:-',self.item_count
```

```python
# Create object of Stack class
obj=Stack()
```

```python
# insert elements in Stack
obj.push(10)
obj.push(20)
obj.push(30)
obj.push(40)
obj.push(50)
```

```python
# top element
obj.peek()
```

```python
# Remove top element
obj.pop()
```

```python
# new top element
obj.peek()
```

```python
# Total number of elements of Stack
obj.size()
```
