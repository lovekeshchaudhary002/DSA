# Circular Singly Linked List

```python
# Define a class Node to describe a node
class Node:
  def __init__(self,item=None,next=None):
    self.item=item
    self.next=next

# Define a class CLL to create and initialize last reference variable
class CLL:
  def __init__(self,last=None):
    self.last=last

  # Define is_empty() method in CLL to check if list is empty
  def is_empty(self):
    return self.last==None

  # Define insert_at_start() method in CLL to insert an element at start of list
  def insert_at_start(self,element):
    n=Node(element)
    if self.is_empty():
      n.next=n
      self.last=n
    else:
      n.next=self.last.next
      self.last.next=n

  # Define insert_at_last() method in CLL to insert an element at last of list
  def insert_at_last(self,element):
    n=Node(element)
    if self.is_empty():
      n.next=n
      self.last=n
    else:
      n.next=self.last.next
      self.last.next=n
      self.last=n

  # Define search() method in CLL to find a specific element in list
  def search(self,value):
    if self.is_empty():
      return None
    else:
      temp=self.last.next
      while temp.next is not self.last:
        if temp.item==value:
          return temp
        temp=temp.next
      if temp.item==value:
        return temp
      return None

  # Define insert_after() method in CLL to insert an element after a specific item
  def insert_after(self,temp,element):
    if temp is not None:
      n=Node(element,temp.next)
      temp.next=n
      if temp==self.last:
        self.last=n

  # Define print_list() method in CLL to print all elements of list
  def print_list(self):
    if self.is_empty():
      print('List is empty')
    else:
      temp=self.last.next
      while temp is not self.last:
        print(temp.item,end=' ')
        temp=temp.next
      print(temp.item)

  # Define delete_first() method in CLL to delete first element
  def delete_first(self):
    if self.is_empty():
      pass
    elif self.last.next==self.last:
      self.last=None
    else:
      self.last.next=self.last.next.next

  # Define delete_last() method in CLL to delete last element
  def delete_last(self):
    if self.is_empty():
      pass
    elif self.last.next==self.last:
      self.last=None
    else:
      temp=self.last.next
      while temp.next is not self.last:
        temp=temp.next
      temp.next=self.last.next
      self.last=temp

  # Define delete_item() method in CLL to delete a specific element
  def delete_item(self,value):
    if self.is_empty():
      pass
    elif self.last.next==self.last:
      if self.last.item==value:
        self.last=None
    else:
      temp=self.last.next
      if temp.item==value:
        self.last.next=temp.next
      else:
        while temp.next is not self.last:
          if temp.next.item==value:
            temp.next=temp.next.next
            break
          temp=temp.next
          if temp.next==self.last:
            if temp.next.item==value:
              temp.next=self.last.next
              self.last=temp
          
```

```python
obj=CLL()
```

```python
# Insert Elements
obj.insert_at_start(30)
obj.insert_at_last(50)
obj.insert_at_start(20)
obj.insert_at_last(70)
obj.insert_after(obj.search(30),40)
obj.insert_after(obj.search(50),60)

obj.print_list()
```
