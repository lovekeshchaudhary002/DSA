# Implementation of Queue by inheritin List class

```python
# Define Queue class to implement queue by inheriting List class
class Queue(list):

  # Define is_empty() method to check if Queue is empty
  def is_empty(self):
    return len(self)==0

  # Define enqueue() method to insert element in Queue (First In)
  def enqueue(self,element):
    self.append(element)

  # Define dequeue() method to remove front element from Queue (First Out)
  def dequeue(self):
    if self.is_empty():
      pass
    else:
      return 'Removed first element:',self.pop(0)

  # Define get_front() method to show first element in Queue
  def get_front(self):
    if self.is_empty():
      print('Queue is empty')
    else:
      return 'front element:',self[0]

  # Define get_rear() method to show last element in Queue
  def get_rear(self):
    if self.is_empty():
      print('Queue is empty')
    else:
      return 'rear element:',self[-1]

  # Define size() method to find number of total elements in Queue
  def size(self):
    return 'Total elements in Queue:',len(self)

  # Restrict insert method of List class by method overriding
  def insert(self,index,element):
    raise AttributeError ('No use of insert method')
```

```python
# Create object of Queue class
obj=Queue()
```

```python
# Insert elements in Queue
obj.enqueue(10)
obj.enqueue(20)
obj.enqueue(30)
obj.enqueue(40)
obj.enqueue(50)
```

```python
# Show front element in Queue
obj.get_front()
```

```python
# Show rear element in Queue
obj.get_rear()
```

```python
# Remove first element from Queue
obj.dequeue()
```
