# Singly Linked List

```python
# Define a class Node to describe a node
class Node:
  def __init__(self,item=None,next=None):
    self.item=item
    self.next=next

# Define a class SLL to create and initialise start reference variable
class SLL:
  def __init__(self,start=None):
    self.start=start

  # Define is_empty() method in SLL to check if the list is empty
  def is_empty(self):
    return self.start==None

  # Define insert_at_start() method in SLL to insert an element in starting
  def insert_at_start(self,element):
    n=Node(element,self.start)
    self.start=n

  # Define insert_at_last() method in SLL to insert an element in last
  def insert_at_last(self,element):
    n=Node(element)
    if self.is_empty():
      self.start=n
    else:
      temp=self.start
      while temp.next is not None:
        temp=temp.next
      temp.next=n

  # Define search() method to find specific node with given item value
  def search(self,value):
    temp=self.start
    while temp.next is not None:
      if temp.item==value:
        return temp
        break
      temp=temp.next
    return None

  # Define insert_after() method in SLL to insert an element after particular node
  def insert_after(self,temp,element):
    if temp is not None:
      n=Node(element,temp.next)
      temp.next=n

  # Define a method in SLL to print all elements
  def display_list(self):
    temp=self.start
    while temp is not None:
      print(temp.item,end=" ")
      temp=temp.next
    print()
    
  # Define a delete_first() method in SLL to delete first element
  def delete_first(self):
    if self.is_empty():
      print("List is empty")
    else:
      self.start=self.start.next

  # Define a delete_last() method in SLL to delete last element
  def delete_last(self):
    if self.is_empty():
      print("List is empty")
    elif self.start.next is None:
      self.start=None
    else:
      temp=self.start
      while temp.next.next is not None:
        temp=temp.next
      temp.next=None

  # Define a delete_item() method in SLL to delete a particular element
  def delete_item(self,element):
    if self.is_empty():
      print('list is empty')
    elif self.start.next==None:
      if self.start.item==element:
        self.start=None
    else:
      temp=self.start
      if temp.item==element:
        self.start=temp.next
      else:
        while temp.next is not None:
          if temp.next.item==element:
            temp.next=temp.next.next
            break
          temp=temp.next
      
```

```python
obj=SLL()
```

```python
# Insert elements in List
obj.insert_at_start(40)
obj.insert_at_start(30)
obj.insert_at_last(70)
obj.insert_at_last(80)
obj.insert_after(obj.search(40),50)
obj.insert_after(obj.search(50),55)
obj.insert_after(obj.search(40),45)
obj.insert_after(obj.search(70),75)
```

```python
print(obj.display_list())
```

```python
obj.delete_first()
obj.delete_last()
obj.delete_item(55)
```
