# Implementation of Dequeue by inheriting List class

```python
# Define Dequeue class to implement Dequeue by inheriting List class
class Dequeue(list):

  # Define is_empty() method to check if Dequeue is empty
  def is_empty(self):
    return len(self)==0

  # Define insert_front() method to insert an element at front end of Dequeue
  def insert_front(self,element):
    self.insert(0,element)

  # Define insert_rear() method to insert an element at rear end of Dequeue
  def insert_rear(self,element):
    self.append(element)

  # Define delete_front() method to remove front element from Dequeue
  def delete_front(self):
    if self.is_empty():
      print("Dequeue is empty")
    else:
      self.pop(0)

  # Define delete_rear() method to remove rear element from Dequeue
  def delete_rear(self):
    if self.is_empty():
      print("Dequeue is empty")
    else:
      self.pop()

  # Define get_front() method to show front element of Dequeue
  def get_front(self):
    if self.is_empty():
      print("Dequeue is empty")
    else:
      return 'Front element',self[0]

  # Definr get_rear() method to show rear element of Dequeue
  def get_rear(self):
    if self.is_empty():
      print("Dequeue is empty")
    else:
      return 'Rear element:',self[-1]

  # Define size() method to find total number of elements in Dequeue
  def size(self):
    return 'Total elements:',len(self)
```

```python
# Create object of Dequeue class
obj=Dequeue()
```

```python
# Insert elements into Dequeue
obj.insert_front(20)
obj.insert_front(15)
obj.insert_rear(25)
obj.insert_rear(30)
obj.insert_front(10)
```

```python
# Number of elements in Dequeue
obj.size()
```

```python
# Show front element in Dequeue
obj.get_front()
```

```python
# Show rear element in Dequeue
obj.get_rear()
```

```python
# Delete elements from Dequeue
obj.delete_front()
obj.delete_rear()
```

```python
# Number of elements in Dequeue
obj.size()
```
