# Stack by inheriting List class

```python
# Define a Stack class by inheriting List class 
# parent class - List , child class - Stack
class Stack(list):

  # Define is_empty() method to check if stack is empty
  def is_empty(self):
    return len(self)==0
  
  # Define push() method to insert elements in stack
  def push(self,element):
    self.append(element)

  # Define pop_item() method to remove top element
  # because pop() method already exists in List class so using pop() method in Stack class will create method overriding 
  def pop_item(self):
    if self.is_empty():
      pass
    else:
      return 'Removed element:-',self.pop()
    
  # Define peek() method to show top element of stack
  def peek(self):
    if self.is_empty():
      print('Stack is empty')
    else:
      return 'Top element:-',self[-1]

  # Define size() method to find number of elements in stack
  def size(self):
    return 'Total elements in stack',len(self)

  # Restrict use of insert method of List class by using method overriding
  # Because insert method can add element at any position but in Stack we only insert in last(top)
  def insert(self,index,element):
    raise AttributeError ('No use of insert method')
```

```python
# Create object of Stack class
obj=Stack()
```

```python
# insert elements in stack
obj.push(10)
obj.push(20)
obj.push(30)
obj.push(40)
```

```python
# Top element
obj.peek()
```

```python
# Remove top element
obj.pop_item()
```

```python
# new top element
obj.peek()
```

```python
# Total elements in stack
obj.size()
```

```python
# check insert method
obj.insert(2,25)
```
