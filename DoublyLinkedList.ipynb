# Doubly Linked List

```python
# Define a class Node to describe a Node of Doubly Linked List
class Node:
  def __init__(self,prev=None,item=None,next=None):
    self.prev=prev
    self.item=item
    self.next=next

# Define a class DLL to create and initialize start reference variable
class DLL:
  def __init__(self,start=None):
    self.start=start

  # Define is_empty() method in DLL to check if the list is empty
  def is_empty(self):
    return self.start==None
    
  # Define insert_at_start() method in DLL to insert an element in starting of list
  def insert_at_start(self,element):
    n=Node(None,element,self.start)
    if self.is_empty():
      self.start=n
    else:
      self.start.prev=n
      self.start=n

  # define insert_at_last() method in DLL to insert an element in end of list
  def insert_at_last(self,element):
    if self.is_empty():
      n=Node(None,element,None)
      self.start=n
    else:
      temp=self.start
      while temp.next is not None:
        temp=temp.next
      n=Node(temp,element,None)
      temp.next=n

  # Define search() method to find a specific element in list
  def search(self,value):
    if self.is_empty():
      print('List is empty')
    elif self.start.next is None:
      if self.start.item==value:
        return self.start.item
    else:
      temp=self.start
      while temp is not None:
        if temp.item==value:
          return temp
          break
        temp=temp.next   

  # Define insert_after() method in DLL to insert an element after specific element in list
  def insert_after(self,temp,element):
    if temp.next is not None:
      n=Node(temp,element,temp.next)
      temp.next.prev=n
      temp.next=n
    else:
      n=Node(temp,element,None)
      temp.next=n  

  # Define delete_first() method in DLL to delete first element from list
  def delete_first(self):
    if self.is_empty():
      pass
    elif self.start.next is None:
      self.start=None
    else:
      self.start=self.start.next
      self.start.prev=None

  # Define delete_last() method in DLL to delete last element from list
  def delete_last(self):
    if self.is_empty():
      pass
    elif self.start.next is None:
      self.start=None
    else:
      temp=self.start
      while temp.next is not None:
        temp=temp.next
      temp.prev.next=None

  # Define delete_item() method in DLL to delete a specific element from list
  def delete_item(self,value):
    if self.is_empty():
      pass
    elif self.start.next is None:
      if self.start.item==value:
        self.start=None
    else:
      temp=self.start
      if temp.item==value:
        self.start=self.start.next
        self.start.prev=None
      else:
        while temp is not None:
          temp=temp.next
          if temp.next is None:
            if temp.item==value:
              temp.prev.next=None
              break
          else:
            if temp.item==value:
              temp.prev.next=temp.next
              temp.next.prev=temp.prev
              break



  # Define print_list() method in DLL to print all elements of list
  def print_list(self):
    if self.is_empty():
      print('List is empty')
    else:
      temp=self.start
      while temp is not None:
        print(temp.item,end=' ')
        temp=temp.next
      print()
  
```

```python
obj=DLL()

obj.insert_at_start(30)
obj.insert_at_start(25)
obj.insert_at_start(20)

obj.insert_at_last(50)
obj.insert_at_last(60)

obj.insert_after(obj.search(30),35)
obj.insert_after(obj.search(35),40)
obj.insert_after(obj.search(60),65)

obj.print_list()
```

```python
obj.delete_first()
obj.print_list()
```
